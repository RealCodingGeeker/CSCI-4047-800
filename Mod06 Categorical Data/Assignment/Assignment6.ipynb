To measure the psychological state of Americans, a psychology researcher (in 1929) devised
three questions put to 3443 Americans from different walks of life. He asked them whether they
were happy. They could answer only **yes** or **no** to each question. By today’s standards the study
is quite simple, and he did no formal statistical analysis of his data. The frequency data (number
of **yes** and **no** responses within each employment category) from this study are in the file
*Americandata.csv*.

---

You need to use the data and answer the question, that is, **Is the type of profession related to the happiness of American people?**
Report the **χ2** statistic and the *statistical significance of test*, in addition to the contingency table.
Also, explain the results and what they mean. What does the value of **χ2** statistic mean? Is it
good or bad? Is the test statistically significant?

```python
import pandas as pd
import scipy.stats as stats


df = pd.read_csv("./Americandata.csv")

# Set Happy as the row labels
contingency_table = df.set_index("Happy")

print("Contingency Table:")
print(contingency_table)

# Chi-square test
chi2, p_value, dof, expected = stats.chi2_contingency(
    contingency_table,
    correction=False
)

print("\nChi-square =", chi2)
print("Degrees of freedom =", dof)
print("p-value =", p_value)
```

Put your analysis in the empty `markdown` cell below.

# Analysis

The contingency table shows the number of Yes and No happiness responses for each profession. A chi-square test of independence was performed to determine whether profession and happiness are related.

The chi-square statistic was **936.1395** with **7 degrees of freedom**. The p-value was **7.5234e-198**.

Since the p-value is much smaller than the significance level of 0.05, we then reject the null hypothesis of independence. Which means, the test is statistically significant, and there is strong evidence that the type of profession is related to the happiness of American people.

The chi-square statistic measures how different the observed frequencies are from the frequencies that would be expected if profession and happiness were independent. A large chi-square value means there is a large difference between the observed and expected frequencies. The value of 936.1395 is very large, indicating a substantial difference between the observed and expected counts.

The chi-square statistic is actually not considered inherently good or bad. It is just a measure of the difference between observed and expected values. In this case, the large chi-square value and extremely small p-value provide strong evidence that profession and happiness are not independent. However, this test shows an association between profession and happiness and does not prove that a person's profession causes them to be happy or unhappy.
