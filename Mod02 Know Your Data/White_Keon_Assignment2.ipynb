4. Plot a *histogram* for **prestige**
5. Plot a *histogram* for **education**

```python
plt.hist(df['prestige'], bins=10, alpha=0.5, label='Prestige')
plt.hist(df['education'], bins=10, alpha=0.5, label='Education')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Distribution of Prestige and Education')
plt.legend()
plt.show()
```

6. Plot a *scatter plot* for **income** and **education**
7. Plot a *scatter plot* for **education** and **prestige**

```python
plt.scatter(x=df['income'], y = df['education'], alpha=0.3)
plt.title("Income compared to Education")
plt.xlabel('Income')
plt.ylabel("Education")
plt.show()
```

```python
plt.scatter(x=df['education'], y = df['prestige'], alpha=0.3)
plt.title("Education compared to Prestige")
plt.xlabel('Education')
plt.ylabel("Prestige")
plt.show()
```

8. Calculate and display the *variance* and *standard deviation* for **income**


```python
income_min = np.min(df['income'])
income_q1 = np.percentile(df['income'], 25)
income_median = np.median(df['income'])
income_mean = df['income'].mean()
income_q3 = np.percentile(df['income'], 75)
income_max = np.max(df['income'])
income_std = df['income'].std()
income_variance = income_std ** 2


print("Minimum:", income_min)
print("Q1:", income_q1)
print("Median:", income_median)
print("Mean:", income_mean)
print("Q3:", income_q3)
print("Maximum:", income_max)
print("Standard Deviation:", income_std)
print("Variance:", income_variance)
```
